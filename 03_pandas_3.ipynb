# groupby() 와 query()

```python
# 경고 메시지 무시
import warnings
warnings.filterwarnings(action='ignore') 

import pandas as pd
```

```python
exam = pd.read_csv('data/exam.csv')
exam
```

```python
exam.groupby('nclass')[['math', 'english', 'science']].mean()
```

```python
exam.groupby('nclass').agg(mean_math = ('math', 'mean'),
                         sum_math    = ('math', 'sum'),          
                         median_math = ('math', 'median'),        
                         n_count = ('nclass', 'count')) 
                           
```

```python
exam.groupby('nclass').agg( { 'math' : ['max','min', 'mean'],   
                              'nclass' : 'count'}) 
```

```python
exam.groupby('nclass').agg( { 'math' : ['max','min', 'mean'],  
                              'english' : ['max','min', 'mean'],
                              'nclass' : 'count'}) 
```

```python
exam.query('nclass==1')
```

```python
exam.query('math >= 70')
```

```python
exam.query('nclass==1 and math >= 50')
```

```python
# 1, 3, 5반 데이터 추출
exam.query('nclass==1 or nclass==3 or nclass==5')
```

```python
# 1, 3, 5반 데이터 추출
exam.query('nclass in [1, 3, 5]')
```

```python
# 1반 수학, 과학 점수 평균 구하기
exam.query('nclass == 1')[['math', 'science']].mean()
```

```python
## 변수 이용 query 조건 만들기

a = int(input("몇 반 학생의 데이터 조회를 원하십니까? "))
exam.query('nclass == @a')
```
