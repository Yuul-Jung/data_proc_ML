# 데이터 병합 (merge(), concat())

```python
# 중간고사 데이터 만들기
test1 = pd.DataFrame({'id'      : [1, 2, 3, 4, 5, 7],
                      'midterm' : [60, 80, 70, 90, 85, 20]})

# 기말고사 데이터 만들기
test2 = pd.DataFrame({'id'    : [1, 2, 3, 4, 5, 6],
                      'final' : [70, 83, 65, 95, 80, 10]})

display(test1)
display(test2)
```

```python
# id 기준으로 merge
total = pd.merge( test1, test2, on='id', how='right')
total
```

```python
total = pd.merge( test1, test2, on='id')   # how = 'inner' (공통데이터_교집합)
total
```

```python
total = pd.merge( test1, test2, on='id', how='outer')  # 양쪽 모든 데이터 (합집합)
total
```

```python
group_a = pd.DataFrame({'id'   : [1, 2, 3, 4, 5],
                        'test' : [60, 80, 70, 90, 85]})

group_b = pd.DataFrame({'id'   : [5, 7, 8, 9, 10],
                        'test' : [70, 83, 65, 95, 80]})
```

```python
# concat() 이용하여 합치기

group_all = pd.concat([group_a, group_b], ignore_index=True)
group_all
```

# loc 와 iloc

```python
# 예제 데이터프레임
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'Score': [85, 90, 95]
}
df = pd.DataFrame(data, index=['A', 'B', 'C'])
df
```

```python
df.loc[['A']]  # 행 'A' 조회
```

```python
df.iloc[0, ]
```

```python
df.iloc[:, 0]
```

```python
# Age가 28보다 큰 score 조회
df.loc[df['Age'] > 28 , 'Score']
```
