```python
## 2차원 리스트를 데이터프레임으로 변환

score=[['정약용', 70, 80, 100],
       ['이순신', 100, 80, 70],
       ['이율곡', 90, 95, 90]]
#print(type(score))
score
```

```python
score_df = pd.DataFrame(score)
print(type(score_df))
score_df
```

```python
score_df.columns = ['이름', '국어', '영어', '수학']
score_df
```

```python
#info() : data 구조 확인 : 결측, 데이터 타입 확인
score_df.info()
```

```python
# stat : min, 1/4, 1/2, 3/4, max 확인
score_df.describe()
```

```python
score_df.columns = ['성명', 'kor', 'eng', 'mat']
score_df
```

```python
score_df.index = score_df['성명']
score_df
```

```python
score_df.set_index('성명', drop=True, inplace=True) ## '성명'을 인덱스로 하고, '성명'항목 삭제
score_df
```

```python
score_df.drop(['eng'], axis=1 , inplace=True)  # "eng"항목 삭제
score_df
```

```python
score_df.drop(['이순신'], axis=0 , inplace=True)  # "이순신""행" 삭제
score_df
```

```python
#데이터 타입 변경
score_df1 = score_df.astype({'kor':float, 'mat':str})
score_df1
```
